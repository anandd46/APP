```python
def sumn(n,num=None):
    if num is None:
        num=[float(input(f"Enter number{i+1}:")) for i in range(n)]
    return sum(num)
n=int(input("Enter n: "))
print(sumn(n))
```

```python
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
n=int(input("Enter the num: "))
for i in range(n):
    print(fib(i),end=" ")
```

```python
#an online store has a list of product prices
# The system must:
#-add 18% gst to every product 
#- consider only products costing more than 500
from functools import reduce
pr=list(map(float,input("enter price").split()))
prgst=list(map(lambda x: x*1.18,pr))
filteredpr=list(filter(lambda x: x>500,prgst))
t=reduce(lambda x,y: x+y,filteredpr,0)
print("Original prices",pr)
print("Price with gst",prgst)
print("Poducts above 500:",filteredpr)
print("Total:",round(t,2))
```

```python
#sc 2: student performance analyzer -'args'
#an mca student may study a diffent number of subjects depending on the semester ,the university wants a program that can calculate the students percetage and grade regardless of the number of subjects.
#problem statemnet-develop a function using args that:
# accepts amrks for any number of subjects,-cal total and percetage,-determines the grade,- returns the percentage and grade.

def std(*mark):
    total =sum(mark)
    per=total/len(mark)
    if per>=90:
        grade="A+"
    elif per>=80:
        grade="A"
    elif per>=70:
        grade="B"
    elif per>=60:
        grade="C"
    elif per>=50:
        grade="D" 
    else:
        grade="F"
    print("Total Marks:",total)
    print("percentage:",per,"%")
    print("Grade:",grade)
std(85,90,78,92,88)
```
