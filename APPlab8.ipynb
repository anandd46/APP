```python
class Counter:
    def __init__(self,start,end):
        self.current=start
        self.end=end
    def __iter__(self):
        return self
    def __next__(self):
        if self.current<=self.end:
            value=self.current
            self.current+=1
            return value
        raise StopIteration

Counter1=Counter(1,5)

print("With iterable object using loop")
for num in Counter1:
    print(num)
print("Using next() manually instad of loop")
counter1=Counter(11,15)
next(counter1)
next(counter1)
next(counter1)
next(counter1)
next(counter1)
```

```python
def counter(start,end):
    for i in range(start,end+1):
        yield i
for num in counter(1,5):
    print(num)
```

```python
def readlog(logs):
    for log in logs:
        yield log
def extract_status(logs):
    for log in logs:
        tt,uid,status=log.split(",")
        yield tt,uid,int(status)
def fil_err(records):
    for tt,uid,status in records:
        if status >= 400:
            yield tt,uid,status
logs=["10:01,u101,200","10:02,u103,404","10:03,u103,200","10:04,u104,500","10:05,u105,403"]
data=readlog(logs)
data=extract_status(data)
errors=fil_err(data)
count=0
for error in errors:
    print(error)
    count+=1
print("Total errors:",count)
```

```python
#given a sentence ,count how many times each word occurs
from collections import Counter
sen="Python data science Python"
words=sen.split()
freq=Counter(words)
print("Freq of words:",freq)
print("Display the most common words",freq.most_common(2))
```

```python
from collections import defaultdict
std=[("Anand","cse"),("Rahul","ece"),("ganu","cse")]
depts=defaultdict(list)
for name ,dept in std:
    depts[dept].append(name)
for dept,names in depts.items():
    print(dept,":",names)
```

```python
from collections import deque
stds=deque(["Sid","ullu","gan"])
stds.append("triveni")
print("queue:",stds)

std=stds.popleft()
stds.appendleft("basavanna")

print("processing:",std)
std=stds.pop()
print("Remaining:",stds)
```

```python
# find the product of all numbers in the list
from functools import reduce
num=[2,3,4,5]
result=reduce(lambda x,y:x*y,num)
print("product:",result)
```

```python
#create a function to calculate the final price ####done
from functools import partial
def cal_pr(pr,tax):
    return pr+(pr*tax/100)
gst_18=partial(cal_pr,tax=18)
print("laptop pr:",gst_18(50000))
print("mobile pr:",gst_18(20000))
```

```python
#Euclidean distance
import math
x1,y1=2,3
x2,y2=7,9
d=math.sqrt((x2-x1)**2+(y2-y1)**2)
print("Distance:",d)
```

```python
# Manhattan distance
x1,y1=2,3
x2,y2=7,9
d = abs(x2 - x1) + abs(y2 - y1)
print("Distance:", d)
```

```python
# Minkowski distance
x1, y1=2,3
x2, y2=7,9
p = 3
d = (abs(x2-x1)**p + abs(y2-y1)**p)**(1/p)
print("Distance:", d)
```
