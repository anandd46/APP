```python
#Anand - MY.AC.P2MCA25003
def fibonacci(n, a=0, b=1):
    if n == 0:
        return a
    else:
        return fibonacci(n - 1, b, a + b)

for i in range(6):
    print(fibonacci(i))
```

```python
def print_factorials(n, accumulator=1):
    if n == 0:
        return
    else:
        print(accumulator)
        print_factorials(n - 1, n * accumulator)

print_factorials(5)
```

```python
def factorial(n, accumulator=1):
    if n == 0:
        return accumulator
    else:
        return factorial(n - 1, n * accumulator)

print(factorial(5))
```

```python
def sum_of_n(n, accumulator=0):
    if n == 0:
        return accumulator
    else:
        return sum_of_n(n - 1, n + accumulator)

for i in range(1, 6):
    print(sum_of_n(i))
```
