```python
# Tree Traversal – Preorder, Inorder and Postorder
tree = {
    'a': {'left': 'b', 'right': 'c'},
    'b': {'left': 'd', 'right': 'e'},
    'c': {'left': None, 'right': None},
    'd': {'left': None, 'right': None},
    'e': {'left': None, 'right': None}
}
def preorder(node):
    if node is None:
        return
    print(node, end=" ")
    preorder(tree[node]['left'])
    preorder(tree[node]['right'])
def inorder(node):
    if node is None:
        return
    inorder(tree[node]['left'])
    print(node, end=" ")
    inorder(tree[node]['right'])
def postorder(node):
    if node is None:
        return
    postorder(tree[node]['left'])
    postorder(tree[node]['right'])
    print(node, end=" ")

print("Preorder:")
preorder('a')
print("\nInorder:")
inorder('a')
print("\nPostorder:")
postorder('a')
```

```python
#bfs using dictionary and queue
from collections import deque
def bfs(g, s):
    visit = {s}
    queue = deque([s])
    while queue:
        v = queue.popleft()
        print(v, end=" -> ")
        for n in g[v]:
            if n not in visit:
                visit.add(n)
                queue.append(n)
graph = {'a': ['b', 'c'],'b': ['a', 'd'],'c': ['a', 'e'],'d': ['b'],'e': ['c']}
bfs(graph, 'a')
```

```python
#dfs using dictionary and queue
def dfs(g, s, visit=None):
    if visit is None:
        visit = set()
    visit.add(s)
    print(s, end=" -> ")
    for n in g[s]:
        if n not in visit:
            dfs(g, n, visit)
graph = {'a': ['b', 'c'],'b': ['a', 'd'],'c': ['a', 'e'],'d': ['b'],'e': ['c']}
dfs(graph, 'a')
```
