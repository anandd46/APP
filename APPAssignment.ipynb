## 07 - Minimize Consecutive Zeros

```python
L = 4
K = 2
a = []
b = []
mn = L + 1
def f(s, o):
    global mn
    if len(s) == L:
        if o == K:
            z = s.split('1')
            m = max([len(i) for i in z])
            a.append((s, m))
            mn = min(mn, m)
        return
    if o < K:
        f(s + '1', o + 1)
    if len(s) - o < L - K:
        f(s + '0', o)
f("", 0)
print("Possible Strings:")
for s, m in a:
    print(s, "→ max zeros =", m)
for s, m in a:
    if m == mn:
        b.append(s)
print("Optimal Arrangement(s):", ", ".join(b))
print("Minimum Longest Zero Block:", mn)
```

## 39 - Matrix Chain Multiplication

```python
p = [10,30,5,60]
n = len(p) - 1
dp = [[0] * n for i in range(n)]
sp = [[0] * n for i in range(n)]
for x in range(2, n + 1):
    for i in range(n - x + 1):
        j = i + x - 1
        dp[i][j] = float('inf')
        for k in range(i, j):
            c = dp[i][k] + dp[k + 1][j] + p[i] * p[k + 1] * p[j]
            if c < dp[i][j]:
                dp[i][j] = c
                sp[i][j] = k
def f(i, j):
    if i == j:
        return "A" + str(i + 1)
    k = sp[i][j]
    return "(" + f(i, k) + " × " + f(k + 1, j) + ")"
print("Minimum Cost:", dp[0][n - 1])
print("Optimal Parenthesization:", f(0, n - 1))
```

## 41 - Word Search in Grid

```python
board = [["o","a","a","n"],
         ["e","t","a","e"],
         ["i","h","k","r"],
         ["i","f","l","v"]]
words = ["oath","pea","eat","rain"]
tr = {}
for w in words:
    p = tr
    for c in w:
        if c not in p:
            p[c] = {}
        p = p[c]
    p["#"] = w
m = len(board)
n = len(board[0])
ans = []
def dfs(i,j,p):
    c = board[i][j]
    if c not in p:
        return
    q = p[c]
    if "#" in q:
        ans.append(q["#"])
        del q["#"]
    board[i][j] = "#"
    if i > 0 and board[i-1][j] != "#":
        dfs(i-1,j,q)
    if i < m-1 and board[i+1][j] != "#":
        dfs(i+1,j,q)
    if j > 0 and board[i][j-1] != "#":
        dfs(i,j-1,q)
    if j < n-1 and board[i][j+1] != "#":
        dfs(i,j+1,q)
    board[i][j] = c
for i in range(m):
    for j in range(n):
        dfs(i,j,tr)
print(ans)
```

## 43 - Euler Path

```python
n = 4
e = [[0,1],[1,2],[2,0],[1,3]]
g = [[] for _ in range(n)]
d = [0] * n
for u,v in e:
    g[u].append(v)
    g[v].append(u)
    d[u] += 1
    d[v] += 1
o = []
for i in range(n):
    if d[i] % 2:
        o.append(i)
v = [False] * n
def dfs(x):
    v[x] = True
    for y in g[x]:
        if not v[y]:
            dfs(y)
s = -1
for i in range(n):
    if d[i] > 0:
        s = i
        break
if s != -1:
    dfs(s)
ok = True
for i in range(n):
    if d[i] > 0 and not v[i]:
        ok = False
if len(o) != 0 and len(o) != 2:
    ok = False
print("Euler Path Exists:", str(ok).lower())
print("Node Degrees:")
for i in range(n):
    print(i, "->", d[i])
print("Odd Degree Nodes:", o)
if ok:
    if len(o) == 2:
        s = o[0]
    g2 = [x[:] for x in g]
    st = [s]
    p = []
    while st:
        x = st[-1]
        if g2[x]:
            y = g2[x].pop()
            g2[y].remove(x)
            st.append(y)
        else:
            p.append(st.pop())
    p.reverse()
    print("Start Node:", s)
    print("Euler Path:", " -> ".join(map(str,p)))
```

## 55 - Custom Iterator

```python
class NumberIterator:
    def __init__(self, start, stop, step):
        self.x = start
        self.stop = stop
        self.step = step
    def __iter__(self):
        return self
    def __next__(self):
        if self.x <= self.stop:
            r = self.x
            self.x += self.step
            return r
        raise StopIteration
x = NumberIterator(0.0, 1.0, 0.25)
for i in x:
    print(i)
```

## 62 - Rat in a Maze

```python
maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]
n = len(maze)
sol = [[0] * n for i in range(n)]
def safe(x, y):
    return x >= 0 and x < n and y >= 0 and y < n and maze[x][y] == 1 and sol[x][y] == 0
def path(x, y):
    if x == n - 1 and y == n - 1:
        sol[x][y] = 1
        return True
    if safe(x, y):
        sol[x][y] = 1
        if path(x + 1, y):
            return True
        if path(x, y + 1):
            return True
        if path(x - 1, y):
            return True
        if path(x, y - 1):
            return True
        sol[x][y] = 0
    return False
if path(0, 0):
    print("Solution Path:")
    for i in sol:
        print(*i)
else:
    print("No solution exists")
```
