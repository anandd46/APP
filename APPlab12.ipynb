```python
# calculate the total size of files
import os
tsize=0
for root,dirs,file in os.walk("clg"):
    for f in file:
        path=os.path.join(root,f)
        try:
            tsize+=os.path.getsize(path)
        except OSError:
            print("Could not access:",path)
print("Total size:",tsize,"bytes")
```

```python
#using json.dumps() and json.loads()
import json
std={"name":"Anand","Marks":90,"Passed":True}
jtxt=json.dumps(std)
print("JSON data:",jtxt)
restoredstd=json.loads(jtxt)
print("Std name:",restoredstd["name"])
print("Marks",restoredstd["Marks"])
```

```python
#using json.dumps() and json.loads()
import json
std={"rollno":102,"name":"Anand","Marks":[90,80,79]}
#write to json file
with open("std.txt","w",encoding="utf-8") as f:
    json.dump(std,f,indent=4)
print("Std data saved")
#read from json file
with open("std.txt","r",encoding="utf-8") as f:
    data=json.load(f)
print("Std:",data["name"])
print("Total marks:", sum(data["Marks"]))
```

```python
#Handling invalid json
import json
txt='{"name":"Anand",Marks":90}'
try:
    data=json.loads(txt)
    print(data)
except json.JSONDecodeError as error:
    print("Invalid JSON",error)
```

```python
import csv
filename = "result.csv"
with open(filename, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    students = list(reader)
    fields = reader.fieldnames.copy()
for column in ["total", "avg"]:
    if column not in fields:
        fields.append(column)
for student in students:
    pymark = float(student["pymark"])
    dbmsmark = float(student["dbmsmark"])
    total = pymark + dbmsmark
    avg = total / 2
    student["total"] = total
    student["avg"] = avg
with open(filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fields)
    writer.writeheader()
    writer.writerows(students)
print("Exist CSV file updatedd")
```
