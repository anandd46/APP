```python
sid = input("Enter record id: ").strip()
found = False
with open(r"op.txt","r", encoding="utf-8") as file:
    for line in file:
        line = line.strip()
        rid = line.split(":")[0].split()[1]
        if rid == sid:
            print("Record found:", rid)
            found = True
            break
if not found:
    print("Record not found")
```

```python
import pickle
import os
f = "std_pkl.pkl"
def loadr():
    if os.path.exists(f):
        with open(f, "rb") as file:
            return pickle.load(file)
    return []
def saverecord(record):
    with open(f, "wb") as file:
        pickle.dump(record, file)
def add_std(record):
    rollno = input("Enter roll: ").strip()
    if any(s["rollno"] == rollno for s in record):
        print("Roll number already exists")
        return
    name = input("Enter name: ").strip()
    marks = float(input("Enter marks: "))
    record.append({
        "rollno": rollno,
        "name": name,
        "marks": marks
    })
    saverecord(record)
    print("Student record saved successfully")
def displaystd(record):
    if not record:
        print("No records found")
        return
    for std in record:
        print(std)
def searchstd(record):
    rollno = input("Enter roll number: ").strip()
    for std in record:
        if std["rollno"] == rollno:
            print("Record found:", std)
            return
    print("Student not found")
record = loadr()
while True:
    print("\n1. Add student")
    print("2. Display students")
    print("3. Search student")
    print("4. Exit")
    ch = input("Enter choice: ").strip()
    if ch == "1":
        add_std(record)
    elif ch == "2":
        displaystd(record)
    elif ch == "3":
        searchstd(record)

    elif ch == "4":
        print("Exiting program")
        break

    else:
        print("Invalid choice")
```
