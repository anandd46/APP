```python
#closure
def outer():
    mes = "Hello"
    def inner():
        print(mes)
    return inner
greet = outer()
greet()
print("Inspect closure:", greet)
print("Inspect using __closure__:", greet.__closure__)
print("cell content:", greet.__closure__[0].cell_contents)
```

```python
#nested funtions
def marks(mark):
    def isvalid(mark):
        return 0<=mark<=100
    for m in mark:
        if isvalid(m):
            print(m)
marks([85,90,105,76])
```

```python
#lexical scope
def outer():
    x=100
    def inner():
        x=10
        print(x)
    inner()
outer()
```

```python
#student Marks grade calculator
def gradecal(passmark):
    def cal(marks):
        if marks>=passmark:
            return "pass"
        else:
            return "fail"
    return cal
ugcheck=gradecal(40)
pgcheck=gradecal(50)
print("Anand:",ugcheck(30))
print("r:",pgcheck(85))
```

```python
#why nonlocal is required
def createacc(intialbal):
    bal=intialbal
    def deposit(amt):
        nonlocal bal
        bal=bal+amt
        return bal
    return deposit
acc=createacc(5000)
print(acc(1000))
print(acc(500))
```
