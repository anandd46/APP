```python
#decorator
def make_multiplier(factor):
    def multiplier(number):
        return number*factor
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print("double function: ", double)
print("triple function: ", triple)
print("double(5): ", double(5))
print("triple(5): ", triple(5))
```

```python
#building decorator manually
def decorator (function):
    def wrapper():
        print("Before the function is called.")
        function()
        print("After the function is called.")
    return wrapper  

def display():
    print("Display function ran.")

display = decorator(display)
display()

```

```python
#decorator with function arguments
def decorator(function):
    def wrapper(a,b):
        print("Addition started: ")
        result=function(a,b)
        print("Addition completed: ")
        return result
    return wrapper
@decorator
def add(a,b):
    return a+b

print("Addition result: ", add(5,10))


```

```python
import time

def timer(function):
    def wrapper():
        start = time.time()
        function()
        end = time.time()
        print("Time taken to execute the function:", end - start)
    return wrapper

@timer
def test():
    time.sleep(2)
    print("Function executed")

test()
```

```python
#A decorator function to measure the time taken to execute a function.

import time
def timer(function):
    def wrapper(*args, **kwargs):
        start=time.time()
        result=function(*args, **kwargs)
        end=time.time()
        print("Time taken to execute the function: ", end-start)
        return result
    return wrapper
@timer
def calculate_sum(n):
    total=0
    for i in range(1,n):
        total+=i
    return total
print("Sum of first 10 million numbers: ", calculate_sum(1000000001))
```

```python
#a bank application requires 1.authentication, 2.logging
#create 2 decorators

def authenticate(function):
    def wrapper(user, *args, **kwargs):
        if user == "admin":
            return function(user, *args, **kwargs)
        print("User is not authenticated.")
        return None
    return wrapper

def log(function):
    def wrapper(*args, **kwargs):
        print("Transaction started:", function.__name__)
        result = function(*args, **kwargs)
        print("Transaction completed:", function.__name__)
        return result
    return wrapper

@authenticate
@log
def Transfer(user, amount):
    print("Transfer of amount", amount, "to user", user, "completed.")

Transfer("admin", 1000)
Transfer("user", 1000)
```

```python
#builtin wrap and decorator with arguments
from functools import wraps
def retry(n):
    def decorator(function):
        @wraps(function)
        def wrapper(*args, **kwargs):
            for i in range(n):
                try:
                    return function(*args, **kwargs)
                except Exception as e:
                    print("Exception occurred: ", e)
                    print("Retrying... ", i+1)
            print("All retries failed.")
        return wrapper
    return decorator

@retry(3)
def connect_server():
    print("Connecting to server...")
    raise Exception("Connection failed.")

connect_server()
```
