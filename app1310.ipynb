```python

from contextlib import contextmanager
@contextmanager
def lab(stdname):
    com="Computer_01"
    print("computer allocated:",stdname)
    print("Allocated manager :",com)
    try:
        yield com
    except Exception as e:
        print("Error occured",e)
        print("Exception successfull")
    finally:
        print("Computer released")


#example 1 normal execution
with lab("anand") as com:
    print("anand is using",com)
    print("Executing python program")

#example 2 exception execution
with lab("genu") as com:
    print("ganu is using",com)
    result=10/0
print("program continues normally ")

 

```

```python
#Ducktypeing
class upi:
    def pay(self,amt):
        print("paid",amt,"using upi")

class creditcard:
    def pay(self,amt):
        print("paid",amt,"using creditcard")

class cash:
    def pay(self,amt):
        print("paid",amt,"using cash")

def processpayment(method,amt):
    method.pay(amt)

processpayment(upi(),500)
processpayment(creditcard(),1200)
processpayment(cash(),700)
```
